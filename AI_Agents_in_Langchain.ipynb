```python
from langchain_classic.agents import AgentExecutor
from langgraph.prebuilt import create_react_agent
from langchainhub import Client

hub = Client()
```

```python
import warnings

warnings.filterwarnings(
    "ignore",
    message=r".*datetime\.datetime\.utcnow\(\) is deprecated.*",
    category=DeprecationWarning
)
```

```python
#Step2: Pull the ReAct prompt from Langchain Hub
prompt=hub.pull('hwchase17/react')
```

**Agent & Agent Executor**




Agent Executor orchestrates the **Entire Loop.**

1. Sends inputs and previous mesages to the agent.
2. Gets the next action from agent
3. Executes that tool with provided input.
4. Adds the tool's observation back into the history.
5. Loops again with updatess history until the agent say Final Answer.

```python
#Step3: Create the ReAct agent manually with the pulled prompt
agent= create_react_agent(
    prompt=prompt,
    model=llm,
    tools=[search_tool]
)
```

```python
#Step 4: Wrap it with AgentExecutor
agent_executor = AgentExecutor(
    agent=agent,
    tools=[search_tool],
    verbose=True
)
```

```python
#Step 5: Invoke
# Step 5: Invoke the LangGraph agent
res = agent.invoke({
    "messages": [
        {"role": "user", "content": "3 ways to reach Goa from Delhi"}
    ]
})

# Print the final response
print(res)
```

**What is ReAct**

**ReAct is a design pattern used in AI agents that stands for Reasoning + Acting. It allows a language model (LLM) to interleave internal reasoning (Thought) with external actions (like tool use) in a astructured , multi-step process.**


In a loop,


THOUGHT


ACTION


OBSERVATION
